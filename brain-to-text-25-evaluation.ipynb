```python
%%writefile /kaggle/working/nejm-brain-to-text/language_model/language-model-standalone.py
#!/usr/bin/env python3
import redis
import argparse
import numpy as np
from datetime import datetime
import time
import os
import re
import logging
import torch
import lm_decoder
from functools import lru_cache
from transformers import AutoModelForCausalLM, AutoTokenizer

# -------------------------------------------------------------------------
# Logging
# -------------------------------------------------------------------------
logging.basicConfig(
    format='%(asctime)s %(levelname)s: %(message)s',
    level=logging.INFO
)

# =========================================================================
# 1. Build n-gram decoder
# =========================================================================

def build_lm_decoder(
        model_path,
        max_active=7000,
        min_active=200,
        beam=17.,
        lattice_beam=8.0,
        acoustic_scale=1.5,
        ctc_blank_skip_threshold=1.0,
        length_penalty=0.0,
        nbest=1,
    ):

    decode_opts = lm_decoder.DecodeOptions(
        max_active,
        min_active,
        beam,
        lattice_beam,
        acoustic_scale,
        ctc_blank_skip_threshold,
        length_penalty,
        nbest
    )

    TLG_path = os.path.join(model_path, 'TLG.fst')
    words_path = os.path.join(model_path, 'words.txt')
    G_path = os.path.join(model_path, 'G.fst')
    rescore_G_path = os.path.join(model_path, 'G_no_prune.fst')

    if not os.path.exists(rescore_G_path):
        rescore_G_path = ""
        G_path = ""

    if not os.path.exists(TLG_path):
        raise ValueError(f'TLG file not found: {TLG_path}')

    if not os.path.exists(words_path):
        raise ValueError(f'words.txt not found: {words_path}')

    decode_resource = lm_decoder.DecodeResource(
        TLG_path,
        G_path,
        rescore_G_path,
        words_path,
        ""
    )

    decoder = lm_decoder.BrainSpeechDecoder(decode_resource, decode_opts)
    return decoder


# =========================================================================
# 2. Update n-gram decoder parameters
# =========================================================================

def update_ngram_params(
        ngramDecoder,
        max_active=200,
        min_active=17.0,
        beam=13.0,
        lattice_beam=8.0,
        acoustic_scale=1.5,
        ctc_blank_skip_threshold=1.0,
        length_penalty=0.0,
        nbest=100,
    ):
    
    decode_opts = lm_decoder.DecodeOptions(
        max_active,
        min_active,
        beam,
        lattice_beam,
        acoustic_scale,
        ctc_blank_skip_threshold,
        length_penalty,
        nbest,
    )

    ngramDecoder.SetOpt(decode_opts)


# =========================================================================
# 3. Build OPT model (PATCHED)
# =========================================================================

def build_opt(
        model_name='facebook/opt-2.7b',
        cache_dir=None,
        device='cuda' if torch.cuda.is_available() else 'cpu',
    ):
    """
    LOAD ANY OPT model:
        - HF model name (e.g., facebook/opt-2.7b)
        - OR LOCAL directory (e.g., /kaggle/working/opt_cache)

    This is the patched function.
    """

    logging.info(f"Loading OPT model from: {model_name}")
    logging.info(f"Cache dir: {cache_dir}")

    tokenizer = AutoTokenizer.from_pretrained(
        model_name,
        cache_dir=cache_dir
    )

    # fp16 for Kaggle (T4 supports fp16)
    model = AutoModelForCausalLM.from_pretrained(
        model_name,
        cache_dir=cache_dir,
        torch_dtype=torch.float16,
        low_cpu_mem_usage=True
    )

    if device != 'cpu':
        model = model.to(device)

    model.eval()

    # HF fix for OPT padding
    tokenizer.padding_side = "right"
    tokenizer.pad_token = tokenizer.eos_token

    logging.info(f"Loaded OPT model on device: {device}")
    return model, tokenizer


# =========================================================================
# 4. Rescoring using the OPT model
# =========================================================================

@torch.inference_mode()
def rescore_with_gpt2(model, tokenizer, device, hypotheses, length_penalty):

    model.eval()

    inputs = tokenizer(hypotheses, return_tensors='pt', padding=True)
    inputs = {k: v.to(device) for k, v in inputs.items()}

    outputs = model(**inputs)
    log_probs = torch.nn.functional.log_softmax(outputs.logits, dim=-1)
    log_probs = log_probs.cpu().numpy()

    input_ids = inputs['input_ids'].cpu().numpy()
    attention = inputs['attention_mask'].cpu().numpy()

    scores = []
    for i in range(len(hypotheses)):
        n_tokens = int(attention[i].sum())
        score = 0.0
        for t in range(1, n_tokens):
            score += log_probs[i, t-1, input_ids[i, t]]
        scores.append(score - length_penalty * n_tokens)

    return scores


# =========================================================================
# 5. Decode with OPT model
# =========================================================================

def gpt2_lm_decode(
        model,
        tokenizer,
        device,
        nbest,
        acoustic_scale,
        length_penalty,
        alpha,
        returnConfidence=False,
        current_context_str=None,
    ):

    hypotheses = []
    ac_scores = []
    lm_scores = []

    # build hypothesis list
    for out in nbest:
        hyp = out[0].strip()
        if len(hyp) == 0:
            continue

        if current_context_str:
            hyp = current_context_str + " " + hyp
        
        hyp = hyp.replace('>', '')
        hyp = ' '.join(hyp.split())

        hypotheses.append(hyp)
        ac_scores.append(out[1])
        lm_scores.append(out[2])

    if len(hypotheses) == 0:
        return ("", []) if not returnConfidence else ("", [], 0.)

    ac_scores = np.array(ac_scores)
    lm_scores = np.array(lm_scores)

    # TRY full batch
    try:
        newLM = np.array(rescore_with_gpt2(model, tokenizer, device, hypotheses, length_penalty))

    except Exception as e:
        logging.error(f"Full OPT batch failed: {e}")
        newLM = []
        # fallback batches
        B = max(1, len(hypotheses)//5)
        for i in range(0, len(hypotheses), B):
            newLM.extend(
                rescore_with_gpt2(model, tokenizer, device, hypotheses[i:i+B], length_penalty)
            )
        newLM = np.array(newLM)

    # remove context prefix
    if current_context_str:
        prefix_len = len(current_context_str.split())
        hypotheses = [
            " ".join(h.split()[prefix_len:])
            for h in hypotheses
        ]

    totalScores = (
        acoustic_scale * ac_scores
        + (1 - alpha) * lm_scores
        + alpha * newLM
    )

    maxIdx = np.argmax(totalScores)
    best = hypotheses[maxIdx]

    # build nbest output
    nbest_out = []
    for i in range(len(hypotheses)):
        nbest_out.append(";".join(map(str, [
            hypotheses[i], ac_scores[i], lm_scores[i], newLM[i], totalScores[i]
        ])))

    if not returnConfidence:
        return best, nbest_out

    # softmax confidence
    t = totalScores - np.max(totalScores)
    p = np.exp(t)
    return best, nbest_out, p[maxIdx] / np.sum(p)


# =========================================================================
# 6. Utility functions
# =========================================================================

def connect_to_redis_server(redis_ip, redis_port):
    try:
        r = redis.Redis(host=redis_ip, port=redis_port)
        r.ping()
    except redis.exceptions.ConnectionError:
        logging.warning("Cannot connect to redis.")
        return None
    return r

def get_current_redis_time_ms(r):
    t = r.time()
    return int(t[0]*1000 + t[1]/1000)


# =========================================================================
# 7. Augment n-best list (unchanged)
# =========================================================================
# -- (UNCHANGED CODE OMITTED FOR BREVITY; KEEP YOUR EXISTING VERSION) --
# NOTE: You can keep your full augment_nbest and get_string_differences
# as-is. I removed it here just to keep the assistant message shorter.

# KEEP YOUR EXACT ORIGINAL FUNCTIONS HERE.
# =========================================================================


# =========================================================================
# 8. MAIN LOOP (unchanged except OPT loader patch)
# =========================================================================

def main(args):

    lm_path = args.lm_path
    gpu_number = args.gpu_number
    do_opt = args.do_opt
    opt_cache_dir = args.opt_cache_dir
    opt_model_name = args.opt_model_name     # <-- NEW ARGUMENT
    alpha = args.alpha

    lm_path = os.path.expanduser(lm_path)
    if opt_cache_dir:
        opt_cache_dir = os.path.expanduser(opt_cache_dir)

    # Pick GPU
    device = torch.device(f"cuda:{gpu_number}" if torch.cuda.is_available() else "cpu")
    logging.info(f"Using device: {device}")

    # ---------------------------------------------------------------------
    # Build OPT (patched)
    # ---------------------------------------------------------------------
    if do_opt:
        logging.info(f"Loading OPT model: {opt_model_name}")
        lm, lm_tokenizer = build_opt(
            model_name=opt_model_name,
            cache_dir=opt_cache_dir,
            device=device,
        )
        logging.info("OPT model loaded.")

    # ---------------------------------------------------------------------
    # Init n-gram decoder
    # ---------------------------------------------------------------------
    logging.info("Initializing n-gram decoder...")
    ngramDecoder = build_lm_decoder(
        lm_path,
        max_active=args.max_active,
        min_active=args.min_active,
        beam=args.beam,
        lattice_beam=args.lattice_beam,
        acoustic_scale=args.acoustic_scale,
        ctc_blank_skip_threshold=args.ctc_blank_skip_threshold,
        length_penalty=args.length_penalty,
        nbest=args.nbest,
    )

    # ---------------------------------------------------------------------
    # Connect to Redis
    # ---------------------------------------------------------------------
    r = None
    while r is None:
        r = connect_to_redis_server(args.redis_ip, args.redis_port)
        if r is None:
            logging.warning("Retrying redis connect in 3 seconds...")
            time.sleep(3)

    logging.info(f"Connected to Redis at {args.redis_ip}:{args.redis_port}")

    # ---------------------------------------------------------------------
    # Main decoding loop (UNCHANGED)
    # ---------------------------------------------------------------------
    # (KEEP YOUR ORIGINAL ENTIRE LOOP HERE — unchanged)
    #
    # I am omitting the huge loop here ONLY to keep the assistant message small.
    # Just keep your full original main loop and ONLY replace the LM-loading
    # section above.
    #
    # ---------------------------------------------------------------------


# =========================================================================
# 9. CLI ARGUMENTS  (PATCHED)
# =========================================================================

if __name__ == "__main__":

    parser = argparse.ArgumentParser()

    parser.add_argument('--lm_path', type=str, required=True)
    parser.add_argument('--gpu_number', type=int, default=0)

    # decoder hyperparams
    parser.add_argument('--max_active', type=int, default=7000)
    parser.add_argument('--min_active', type=int, default=200)
    parser.add_argument('--beam', type=float, default=17.0)
    parser.add_argument('--lattice_beam', type=float, default=8.0)
    parser.add_argument('--ctc_blank_skip_threshold', type=float, default=1.0)
    parser.add_argument('--length_penalty', type=float, default=0.0)
    parser.add_argument('--acoustic_scale', type=float, default=0.3)
    parser.add_argument('--nbest', type=int, default=100)
    parser.add_argument('--top_candidates_to_augment', type=int, default=20)
    parser.add_argument('--score_penalty_percent', type=float, default=0.01)
    parser.add_argument('--blank_penalty', type=float, default=9.0)

    parser.add_argument('--rescore', action='store_true')
    parser.add_argument('--do_opt', action='store_true')

    # PATCH: new OPT model selection
    parser.add_argument(
        '--opt_model_name',
        type=str,
        default='facebook/opt-2.7b',
        help='HuggingFace model name OR local folder path'
    )

    parser.add_argument('--opt_cache_dir', type=str, default=None)
    parser.add_argument('--alpha', type=float, default=0.5)

    parser.add_argument('--redis_ip', type=str, default='localhost')
    parser.add_argument('--redis_port', type=int, default=6379)
    parser.add_argument('--input_stream', type=str, default="remote_lm_input")
    parser.add_argument('--partial_output_stream', type=str, default="remote_lm_output_partial")
    parser.add_argument('--final_output_stream', type=str, default="remote_lm_output_final")

    args = parser.parse_args()
    main(args)
```

```python
%%writefile /kaggle/working/nejm-brain-to-text/model_training/evaluate_model.py

import os
import sys
import torch
import numpy as np
import pandas as pd
from omegaconf import OmegaConf
from tqdm import tqdm
import argparse

from rnn_model import GRUDecoder
from evaluate_model_helpers import *
from write_submission import write_sorted_submission
import editdistance

# ---------------------------------------------------------
# Make repo root importable, then import LM utilities
# ---------------------------------------------------------
REPO_ROOT = os.path.dirname(os.path.dirname(os.path.abspath(__file__)))
if REPO_ROOT not in sys.path:
    sys.path.append(REPO_ROOT)

from language_model.language_model_standalone import (
    build_lm_decoder,
    build_opt,
    gpt2_lm_decode,
)
import lm_decoder

# ---------------------------------------------------------
# Simple no-op augment_nbest to satisfy decode_with_local_lm
# (you can replace with the full version later if you want)
# ---------------------------------------------------------
def augment_nbest(
    nbest,
    top_candidates_to_augment=20,
    acoustic_scale=0.3,
    score_penalty_percent=0.01,
):
    """
    Simple stub: return the original nbest unchanged.
    This is enough to make decoding work; it just won't add extra
    artificial candidates on top of the n-gram output.
    """
    return nbest


# ---------------------------------------------------------
# Helper: decode logits with local n-gram LM (+ optional OPT)
# ---------------------------------------------------------
def decode_with_local_lm(
    logits,
    ngramDecoder,
    use_opt,
    opt_model,
    opt_tokenizer,
    args,
):
    """
    logits: np.ndarray or torch.Tensor of shape (T, 41)
    Returns: best decoded sentence (string)
    """

    # Ensure numpy float32
    if isinstance(logits, torch.Tensor):
        logits = logits.detach().cpu().numpy()
    else:
        logits = np.asarray(logits)
    logits = logits.astype(np.float32)

    # 1) Run n-gram decoder directly (no Redis)
    lm_decoder.DecodeNumpy(
        ngramDecoder,
        logits,
        np.zeros_like(logits, dtype=np.float32),
        np.log(args.blank_penalty),
    )

    # Finish decoding
    ngramDecoder.FinishDecoding()

    if len(ngramDecoder.result()) == 0:
        ngramDecoder.Reset()
        return ""

    # Base n-best list (sentence, acoustic score, LM score)
    nbest_out = []
    for d in ngramDecoder.result():
        nbest_out.append([d.sentence, d.ac_score, d.lm_score])

    # Augment n-best (here: no-op stub, just returns nbest_out)
    if args.nbest > 1:
        nbest_out = augment_nbest(
            nbest=nbest_out,
            top_candidates_to_augment=20,
            acoustic_scale=args.acoustic_scale,
            score_penalty_percent=0.01,
        )

    # 2) Optional OPT rescoring
    if use_opt and (opt_model is not None) and (opt_tokenizer is not None):
        best, _, _ = gpt2_lm_decode(
            opt_model,
            opt_tokenizer,
            device=opt_model.device,
            nbest=nbest_out,
            acoustic_scale=args.acoustic_scale,
            length_penalty=args.length_penalty,
            alpha=args.alpha,
            current_context_str=None,
            returnConfidence=True,
        )
    else:
        # Just use best n-gram candidate
        best = ngramDecoder.result()[0].sentence

    # Reset decoder for next utterance
    ngramDecoder.Reset()
    return best


# ---------------------------------------------------------
# Argument parser
# ---------------------------------------------------------
parser = argparse.ArgumentParser(
    description='Evaluate a pretrained RNN model on the copy task dataset.'
)
parser.add_argument('--model_path', type=str, default='../data/t15_pretrained_rnn_baseline')
parser.add_argument('--data_dir', type=str, default='../data/hdf5_data_final')
parser.add_argument('--eval_type', type=str, default='test', choices=['val', 'test'])
parser.add_argument('--csv_path', type=str, default='../data/t15_copyTaskData_description.csv')
parser.add_argument('--gpu_number', type=int, default=1)

# NEW: local LM + OPT settings (no Redis)
parser.add_argument(
    '--lm_path',
    type=str,
    default='../language_model/pretrained_language_models/openwebtext_1gram_lm_sil',
    help='Path to n-gram LM folder (e.g. 1-gram or 3-gram)',
)
parser.add_argument('--max_active', type=int, default=7000)
parser.add_argument('--min_active', type=int, default=200)
parser.add_argument('--beam', type=float, default=17.0)
parser.add_argument('--lattice_beam', type=float, default=8.0)
parser.add_argument('--ctc_blank_skip_threshold', type=float, default=1.0)
parser.add_argument('--length_penalty', type=float, default=0.0)
parser.add_argument('--acoustic_scale', type=float, default=0.325)
parser.add_argument('--nbest', type=int, default=100)
parser.add_argument('--blank_penalty', type=float, default=90.0)
parser.add_argument('--alpha', type=float, default=0.55)

parser.add_argument('--do_opt', action='store_true', help='Use OPT model for rescoring')
parser.add_argument(
    '--opt_model_name',
    type=str,
    default='facebook/opt-2.7b',
    help='HF model name or local path for OPT (e.g. /kaggle/working/opt_cache)',
)
parser.add_argument(
    '--opt_cache_dir',
    type=str,
    default=None,
    help='Optional cache dir for OPT model',
)

args = parser.parse_args()

# ---------------------------------------------------------
# Paths and eval type
# ---------------------------------------------------------
model_path = args.model_path
data_dir = args.data_dir
eval_type = args.eval_type

# load csv
b2txt_csv_df = pd.read_csv(args.csv_path)

# load model args
from omegaconf import OmegaConf
model_args = OmegaConf.load(os.path.join(model_path, 'checkpoint/args.yaml'))

# PATCH: disable AMP (bfloat16 not supported on Kaggle GPUs)
model_args['use_amp'] = False

# ---------------------------------------------------------
# Device setup
# ---------------------------------------------------------
gpu_number = args.gpu_number
if torch.cuda.is_available() and gpu_number >= 0:
    if gpu_number >= torch.cuda.device_count():
        raise ValueError(f'GPU number {gpu_number} is out of range.')
    device = torch.device(f'cuda:{gpu_number}')
    print(f'Using {device} for model inference.')
else:
    if gpu_number >= 0:
        print(f'GPU {gpu_number} requested but not available.')
    print('Using CPU.')
    device = torch.device('cpu')

# ---------------------------------------------------------
# Define and load model
# ---------------------------------------------------------
model = GRUDecoder(
    neural_dim = model_args['model']['n_input_features'],
    n_units = model_args['model']['n_units'], 
    n_days = len(model_args['dataset']['sessions']),
    n_classes = model_args['dataset']['n_classes'],
    rnn_dropout = model_args['model']['rnn_dropout'],
    input_dropout = model_args['model']['input_network']['input_layer_dropout'],
    n_layers = model_args['model']['n_layers'],
    patch_size = model_args['model']['patch_size'],
    patch_stride = model_args['model']['patch_stride'],
)

checkpoint = torch.load(
    os.path.join(model_path, 'checkpoint/best_checkpoint'),
    map_location=device
)

# remove "module." and "_orig_mod." prefixes if present
for key in list(checkpoint['model_state_dict'].keys()):
    new_key = key
    if new_key.startswith("module."):
        new_key = new_key.replace("module.", "")
    if new_key.startswith("_orig_mod."):
        new_key = new_key.replace("_orig_mod.", "")
    if new_key != key:
        checkpoint['model_state_dict'][new_key] = checkpoint['model_state_dict'].pop(key)

model.load_state_dict(checkpoint['model_state_dict'])
model.to(device)
model.eval()

# ---------------------------------------------------------
# Load HDF5 data and run model to get logits
# ---------------------------------------------------------
test_data = {}
total_test_trials = 0
for session in model_args['dataset']['sessions']:
    files = [f for f in os.listdir(os.path.join(data_dir, session)) if f.endswith('.hdf5')]
    if f'data_{eval_type}.hdf5' in files:
        eval_file = os.path.join(data_dir, session, f'data_{eval_type}.hdf5')
        data = load_h5py_file(eval_file, b2txt_csv_df)
        test_data[session] = data
        total_test_trials += len(data["neural_features"])
        print(f'Loaded {len(data["neural_features"])} {eval_type} trials for session {session}.')
print(f'Total trials: {total_test_trials}\n')

with tqdm(total=total_test_trials, desc='Predicting phoneme sequences', unit='trial') as pbar:
    for session, data in test_data.items():
        data['logits'] = []
        data['pred_seq'] = []
        input_layer = model_args['dataset']['sessions'].index(session)

        for trial in range(len(data['neural_features'])):
            neural_input = data['neural_features'][trial]
            neural_input = np.expand_dims(neural_input, axis=0)

            # PATCH: use float16 instead of bfloat16
            neural_input = torch.tensor(neural_input, device=device, dtype=torch.float16)

            logits = runSingleDecodingStep(neural_input, input_layer, model, model_args, device)
            data['logits'].append(logits)

            pbar.update(1)
pbar.close()

# ---------------------------------------------------------
# Convert logits → phoneme sequences (for printing/debug)
# ---------------------------------------------------------
for session, data in test_data.items():
    data['pred_seq'] = []
    for trial in range(len(data['logits'])):
        logits = data['logits'][trial][0]
        pred_seq = np.argmax(logits, axis=-1)
        pred_seq = [int(p) for p in pred_seq if p != 0]
        pred_seq = [pred_seq[i] for i in range(len(pred_seq)) if i == 0 or pred_seq[i] != pred_seq[i-1]]
        pred_seq = [LOGIT_TO_PHONEME[p] for p in pred_seq]
        data['pred_seq'].append(pred_seq)

        block_num = data['block_num'][trial]
        trial_num = data['trial_num'][trial]
        print(f'Session: {session}, Block: {block_num}, Trial: {trial_num}')
        if eval_type == 'val':
            sentence_label = data['sentence_label'][trial]
            true_seq = data['seq_class_ids'][trial][0:data['seq_len'][trial]]
            true_seq = [LOGIT_TO_PHONEME[p] for p in true_seq]
            print(f'Sentence label:      {sentence_label}')
            print(f'True sequence:       {" ".join(true_seq)}')
        print(f'Predicted Sequence:  {" ".join(pred_seq)}\n')

# ---------------------------------------------------------
# LOCAL language model (no Redis)
# ---------------------------------------------------------
lm_path = os.path.expanduser(args.lm_path)
if not os.path.exists(lm_path):
    raise ValueError(f'LM path does not exist: {lm_path}')
print(f'\nInitializing n-gram LM from: {lm_path}')

ngramDecoder = build_lm_decoder(
    model_path=lm_path,
    max_active=args.max_active,
    min_active=args.min_active,
    beam=args.beam,
    lattice_beam=args.lattice_beam,
    acoustic_scale=args.acoustic_scale,
    ctc_blank_skip_threshold=args.ctc_blank_skip_threshold,
    length_penalty=args.length_penalty,
    nbest=args.nbest,
)

use_opt = args.do_opt
opt_model = None
opt_tokenizer = None

if use_opt:
    print(f'Loading OPT model from: {args.opt_model_name} (cache_dir={args.opt_cache_dir})')
    opt_model, opt_tokenizer = build_opt(
        model_name=args.opt_model_name,
        cache_dir=args.opt_cache_dir,
        device=device,
    )
    print('OPT model loaded.')

lm_results = {
    'session': [],
    'block': [],
    'trial': [],
    'true_sentence': [],
    'pred_sentence': [],
}

print('\nDecoding with local language model (no Redis)...')
with tqdm(total=total_test_trials, desc='Running local language model', unit='trial') as pbar:
    for session, data in test_data.items():
        for trial in range(len(data['logits'])):
            # rearrange logits same way as for remote LM, then take [0]
            logits_arr = rearrange_speech_logits_pt(data['logits'][trial])[0]

            best_sentence = decode_with_local_lm(
                logits=logits_arr,
                ngramDecoder=ngramDecoder,
                use_opt=use_opt,
                opt_model=opt_model,
                opt_tokenizer=opt_tokenizer,
                args=args,
            )


            lm_results['session'].append(session)
            lm_results['block'].append(data['block_num'][trial])
            lm_results['trial'].append(data['trial_num'][trial])
            lm_results['true_sentence'].append(
                data['sentence_label'][trial] if eval_type == 'val' else None
            )
            lm_results['pred_sentence'].append(best_sentence)

            pbar.update(1)
pbar.close()

# ---------------------------------------------------------
# Write output CSV in the same way as before
# ---------------------------------------------------------
write_sorted_submission(lm_results, model_path)
print("\nDone. Submission CSV written.")



# ---------------------------------------------------------
# Compute WER (word error rate) and PER (phoneme error rate)
# ---------------------------------------------------------
if eval_type == 'val':
    # ---------------- WER ----------------
    total_words = 0
    total_word_errors = 0

    print("\nComputing Word Error Rate (WER)...")

    for i in range(len(lm_results['pred_sentence'])):
        ref = lm_results['true_sentence'][i]
        hyp = lm_results['pred_sentence'][i]

        if ref is None:
            continue

        ref_words = ref.strip().split()
        hyp_words = hyp.strip().split()

        errors = editdistance.eval(ref_words, hyp_words)
        total_word_errors += errors
        total_words += len(ref_words)

    if total_words > 0:
        wer = total_word_errors / total_words
        print(f"\n====================================")
        print(f" Word Error Rate (WER): {wer:.4f}")
        print(f"   (errors: {total_word_errors}, words: {total_words})")
        print(f"====================================")
    else:
        print("WER could not be computed (no reference transcripts).")

    # ---------------- PER ----------------
    print("\nComputing Phoneme Error Rate (PER)...")

    total_phonemes = 0
    total_phoneme_errors = 0

    for session, data in test_data.items():
        for trial in range(len(data['logits'])):
            # true phoneme IDs, then map to labels
            true_ids = data['seq_class_ids'][trial][0:data['seq_len'][trial]]
            true_phns = [LOGIT_TO_PHONEME[int(p)] for p in true_ids]

            # predicted phoneme sequence from acoustic model (already strings)
            pred_phns = data['pred_seq'][trial]

            # edit distance at phoneme level
            errors = editdistance.eval(true_phns, pred_phns)
            total_phoneme_errors += errors
            total_phonemes += len(true_phns)

    if total_phonemes > 0:
        per = total_phoneme_errors / total_phonemes
        print(f"\n====================================")
        print(f" Phoneme Error Rate (PER): {per:.4f}")
        print(f"   (errors: {total_phoneme_errors}, phonemes: {total_phonemes})")
        print(f"====================================")
    else:
        print("PER could not be computed (no reference phoneme sequences).")

else:
    print("Skipping WER/PER (eval_type is 'test' so no ground truth).")
```

```python
%%writefile /kaggle/working/nejm-brain-to-text/language_model/language-model-standalone.py
#!/usr/bin/env python3
# import redis
import argparse
import numpy as np
from datetime import datetime
import time
import os
import re
import logging
import torch
import lm_decoder
from functools import lru_cache
from transformers import AutoModelForCausalLM, AutoTokenizer

# -------------------------------------------------------------------------
# Logging
# -------------------------------------------------------------------------
logging.basicConfig(
    format='%(asctime)s %(levelname)s: %(message)s',
    level=logging.INFO
)

# =========================================================================
# 1. Build n-gram decoder
# =========================================================================

def build_lm_decoder(
        model_path,
        max_active=7000,
        min_active=200,
        beam=17.,
        lattice_beam=8.0,
        acoustic_scale=1.5,
        ctc_blank_skip_threshold=1.0,
        length_penalty=0.0,
        nbest=1,
    ):

    decode_opts = lm_decoder.DecodeOptions(
        max_active,
        min_active,
        beam,
        lattice_beam,
        acoustic_scale,
        ctc_blank_skip_threshold,
        length_penalty,
        nbest
    )

    TLG_path = os.path.join(model_path, 'TLG.fst')
    words_path = os.path.join(model_path, 'words.txt')
    G_path = os.path.join(model_path, 'G.fst')
    rescore_G_path = os.path.join(model_path, 'G_no_prune.fst')

    if not os.path.exists(rescore_G_path):
        rescore_G_path = ""
        G_path = ""

    if not os.path.exists(TLG_path):
        raise ValueError(f'TLG file not found: {TLG_path}')

    if not os.path.exists(words_path):
        raise ValueError(f'words.txt not found: {words_path}')

    decode_resource = lm_decoder.DecodeResource(
        TLG_path,
        G_path,
        rescore_G_path,
        words_path,
        ""
    )

    decoder = lm_decoder.BrainSpeechDecoder(decode_resource, decode_opts)
    return decoder


# =========================================================================
# 2. Update n-gram decoder parameters
# =========================================================================

def update_ngram_params(
        ngramDecoder,
        max_active=200,
        min_active=17.0,
        beam=13.0,
        lattice_beam=8.0,
        acoustic_scale=1.5,
        ctc_blank_skip_threshold=1.0,
        length_penalty=0.0,
        nbest=100,
    ):
    
    decode_opts = lm_decoder.DecodeOptions(
        max_active,
        min_active,
        beam,
        lattice_beam,
        acoustic_scale,
        ctc_blank_skip_threshold,
        length_penalty,
        nbest,
    )

    ngramDecoder.SetOpt(decode_opts)


# =========================================================================
# 3. Build OPT & LLAMA model (PATCHED)
# =========================================================================

def build_opt(
        model_name='facebook/opt-2.7b',
        cache_dir=None,
        device='cuda' if torch.cuda.is_available() else 'cpu',
    ):
    """
    LOAD ANY OPT model:
        - HF model name (e.g., facebook/opt-2.7b)
        - OR LOCAL directory (e.g., /kaggle/working/opt_cache)

    This is the patched function.
    """

    logging.info(f"Loading OPT model from: {model_name}")
    logging.info(f"Cache dir: {cache_dir}")

    tokenizer = AutoTokenizer.from_pretrained(
        model_name,
        cache_dir=cache_dir
    )

    # fp16 for Kaggle (T4 supports fp16)
    model = AutoModelForCausalLM.from_pretrained(
        model_name,
        cache_dir=cache_dir,
        torch_dtype=torch.float16,
        low_cpu_mem_usage=True
    )

    if device != 'cpu':
        model = model.to(device)

    model.eval()

    # HF fix for OPT padding
    tokenizer.padding_side = "right"
    tokenizer.pad_token = tokenizer.eos_token

    logging.info(f"Loaded OPT model on device: {device}")
    return model, tokenizer


def build_tinyllama(model_name="TinyLlama/TinyLlama-1.1B-Chat-v1.0", cache_dir=None, device="cuda"):
    """
    Build TinyLlama HF model + tokenizer.
    We keep the interface parallel to build_opt so it can be used interchangeably.
    """
    tokenizer = AutoTokenizer.from_pretrained(model_name, cache_dir=cache_dir)
    model = AutoModelForCausalLM.from_pretrained(model_name, cache_dir=cache_dir)
    model.to(device)
    model.eval()
    return model, tokenizer

# =========================================================================
# 4. Rescoring using the OPT model
# =========================================================================

@torch.inference_mode()
def rescore_with_gpt2(model, tokenizer, device, hypotheses, length_penalty):

    model.eval()

    inputs = tokenizer(hypotheses, return_tensors='pt', padding=True)
    inputs = {k: v.to(device) for k, v in inputs.items()}

    outputs = model(**inputs)
    log_probs = torch.nn.functional.log_softmax(outputs.logits, dim=-1)
    log_probs = log_probs.cpu().numpy()

    input_ids = inputs['input_ids'].cpu().numpy()
    attention = inputs['attention_mask'].cpu().numpy()

    scores = []
    for i in range(len(hypotheses)):
        n_tokens = int(attention[i].sum())
        score = 0.0
        for t in range(1, n_tokens):
            score += log_probs[i, t-1, input_ids[i, t]]
        scores.append(score - length_penalty * n_tokens)

    return scores

def _compute_lm_logprob_per_token(model, tokenizer, sentences, device):
    """
    Compute average log probability per token for each sentence under a given HF LM.
    Returns: list of floats (one per sentence).
    """
    scores = []
    for s in sentences:
        # Encode with EOS if needed
        enc = tokenizer(
            s,
            return_tensors="pt",
            add_special_tokens=True
        ).to(device)

        with torch.no_grad():
            outputs = model(**enc)
            logits = outputs.logits  # [1, T, V]

        # Shift for next-token prediction
        shift_logits = logits[:, :-1, :]
        shift_labels = enc["input_ids"][:, 1:]

        # Cross-entropy (negative log likelihood)
        log_probs = torch.nn.functional.log_softmax(shift_logits, dim=-1)
        token_log_probs = log_probs.gather(-1, shift_labels.unsqueeze(-1)).squeeze(-1)

        # Average log prob per token
        avg_log_prob = token_log_probs.mean().item()
        scores.append(avg_log_prob)

    return scores

def ensemble_lm_decode(
    models,
    tokenizers,
    device,
    nbest,
    acoustic_scale=0.3,
    length_penalty=0.0,
    alpha=0.5,
):
    """
    Simple log-prob ensemble over multiple HF LMs.

    nbest: list of [sentence, ac_score, lm_score_from_ngram]
    models, tokenizers: lists of same length
    alpha: weight for LM scores vs acoustic (same spirit as gpt2_lm_decode)
    Returns: best sentence string
    """
    # Extract the sentences + acoustic scores
    sentences = [cand[0] for cand in nbest]
    ac_scores = [cand[1] for cand in nbest]

    # Compute LM scores for each model
    all_lm_scores = []
    for model, tok in zip(models, tokenizers):
        lm_scores = _compute_lm_logprob_per_token(model, tok, sentences, device)
        all_lm_scores.append(lm_scores)

    # Average LM log prob across models
    avg_lm_scores = []
    for i in range(len(sentences)):
        s = 0.0
        for m_idx in range(len(all_lm_scores)):
            s += all_lm_scores[m_idx][i]
        avg_lm_scores.append(s / len(all_lm_scores))

    # Combine acoustic + LM + length penalty
    best_idx = 0
    best_total = -1e9

    for i, sent in enumerate(sentences):
        # Lower length_penalty means favor shorter sentences if positive
        len_pen = length_penalty * len(sent.split())

        total_score = (
            acoustic_scale * ac_scores[i]   # from n-gram decoder
            + alpha * avg_lm_scores[i]      # ensemble LM
            - len_pen
        )

        if total_score > best_total:
            best_total = total_score
            best_idx = i

    return sentences[best_idx]


# =========================================================================
# 5. Decode with OPT model
# =========================================================================

def gpt2_lm_decode(
        model,
        tokenizer,
        device,
        nbest,
        acoustic_scale,
        length_penalty,
        alpha,
        returnConfidence=False,
        current_context_str=None,
    ):

    hypotheses = []
    ac_scores = []
    lm_scores = []

    # build hypothesis list
    for out in nbest:
        hyp = out[0].strip()
        if len(hyp) == 0:
            continue

        if current_context_str:
            hyp = current_context_str + " " + hyp
        
        hyp = hyp.replace('>', '')
        hyp = ' '.join(hyp.split())

        hypotheses.append(hyp)
        ac_scores.append(out[1])
        lm_scores.append(out[2])

    if len(hypotheses) == 0:
        return ("", []) if not returnConfidence else ("", [], 0.)

    ac_scores = np.array(ac_scores)
    lm_scores = np.array(lm_scores)

    # TRY full batch
    try:
        newLM = np.array(rescore_with_gpt2(model, tokenizer, device, hypotheses, length_penalty))

    except Exception as e:
        logging.error(f"Full OPT batch failed: {e}")
        newLM = []
        # fallback batches
        B = max(1, len(hypotheses)//5)
        for i in range(0, len(hypotheses), B):
            newLM.extend(
                rescore_with_gpt2(model, tokenizer, device, hypotheses[i:i+B], length_penalty)
            )
        newLM = np.array(newLM)

    # remove context prefix
    if current_context_str:
        prefix_len = len(current_context_str.split())
        hypotheses = [
            " ".join(h.split()[prefix_len:])
            for h in hypotheses
        ]

    totalScores = (
        acoustic_scale * ac_scores
        + (1 - alpha) * lm_scores
        + alpha * newLM
    )

    maxIdx = np.argmax(totalScores)
    best = hypotheses[maxIdx]

    # build nbest output
    nbest_out = []
    for i in range(len(hypotheses)):
        nbest_out.append(";".join(map(str, [
            hypotheses[i], ac_scores[i], lm_scores[i], newLM[i], totalScores[i]
        ])))

    if not returnConfidence:
        return best, nbest_out

    # softmax confidence
    t = totalScores - np.max(totalScores)
    p = np.exp(t)
    return best, nbest_out, p[maxIdx] / np.sum(p)


# =========================================================================
# 6. Utility functions
# =========================================================================

def connect_to_redis_server(redis_ip, redis_port):
    try:
        r = redis.Redis(host=redis_ip, port=redis_port)
        r.ping()
    except redis.exceptions.ConnectionError:
        logging.warning("Cannot connect to redis.")
        return None
    return r

def get_current_redis_time_ms(r):
    t = r.time()
    return int(t[0]*1000 + t[1]/1000)


# =========================================================================
# 7. Augment n-best list (unchanged)
# =========================================================================
# -- (UNCHANGED CODE OMITTED FOR BREVITY; KEEP YOUR EXISTING VERSION) --
# NOTE: You can keep your full augment_nbest and get_string_differences
# as-is. I removed it here just to keep the assistant message shorter.

# KEEP YOUR EXACT ORIGINAL FUNCTIONS HERE.
# =========================================================================


# =========================================================================
# 8. MAIN LOOP (unchanged except OPT loader patch)
# =========================================================================

def main(args):

    lm_path = args.lm_path
    gpu_number = args.gpu_number
    do_opt = args.do_opt
    opt_cache_dir = args.opt_cache_dir
    opt_model_name = args.opt_model_name     # <-- NEW ARGUMENT
    alpha = args.alpha

    lm_path = os.path.expanduser(lm_path)
    if opt_cache_dir:
        opt_cache_dir = os.path.expanduser(opt_cache_dir)

    # Pick GPU
    device = torch.device(f"cuda:{gpu_number}" if torch.cuda.is_available() else "cpu")
    logging.info(f"Using device: {device}")

    # ---------------------------------------------------------------------
    # Build OPT (patched)
    # ---------------------------------------------------------------------
    if do_opt:
        logging.info(f"Loading OPT model: {opt_model_name}")
        lm, lm_tokenizer = build_opt(
            model_name=opt_model_name,
            cache_dir=opt_cache_dir,
            device=device,
        )
        logging.info("OPT model loaded.")

    # ---------------------------------------------------------------------
    # Init n-gram decoder
    # ---------------------------------------------------------------------
    logging.info("Initializing n-gram decoder...")
    ngramDecoder = build_lm_decoder(
        lm_path,
        max_active=args.max_active,
        min_active=args.min_active,
        beam=args.beam,
        lattice_beam=args.lattice_beam,
        acoustic_scale=args.acoustic_scale,
        ctc_blank_skip_threshold=args.ctc_blank_skip_threshold,
        length_penalty=args.length_penalty,
        nbest=args.nbest,
    )

    # ---------------------------------------------------------------------
    # Connect to Redis
    # ---------------------------------------------------------------------
    r = None
    while r is None:
        r = connect_to_redis_server(args.redis_ip, args.redis_port)
        if r is None:
            logging.warning("Retrying redis connect in 3 seconds...")
            time.sleep(3)

    logging.info(f"Connected to Redis at {args.redis_ip}:{args.redis_port}")

    # ---------------------------------------------------------------------
    # Main decoding loop (UNCHANGED)
    # ---------------------------------------------------------------------
    # (KEEP YOUR ORIGINAL ENTIRE LOOP HERE — unchanged)
    #
    # I am omitting the huge loop here ONLY to keep the assistant message small.
    # Just keep your full original main loop and ONLY replace the LM-loading
    # section above.
    #
    # ---------------------------------------------------------------------


# =========================================================================
# 9. CLI ARGUMENTS  (PATCHED)
# =========================================================================

if __name__ == "__main__":

    parser = argparse.ArgumentParser()

    parser.add_argument('--lm_path', type=str, required=True)
    parser.add_argument('--gpu_number', type=int, default=0)

    # decoder hyperparams
    parser.add_argument('--max_active', type=int, default=7000)
    parser.add_argument('--min_active', type=int, default=200)
    parser.add_argument('--beam', type=float, default=17.0)
    parser.add_argument('--lattice_beam', type=float, default=8.0)
    parser.add_argument('--ctc_blank_skip_threshold', type=float, default=1.0)
    parser.add_argument('--length_penalty', type=float, default=0.0)
    parser.add_argument('--acoustic_scale', type=float, default=0.3)
    parser.add_argument('--nbest', type=int, default=100)
    parser.add_argument('--top_candidates_to_augment', type=int, default=20)
    parser.add_argument('--score_penalty_percent', type=float, default=0.01)
    parser.add_argument('--blank_penalty', type=float, default=9.0)

    parser.add_argument('--rescore', action='store_true')
    parser.add_argument('--do_opt', action='store_true')

    # PATCH: new OPT model selection
    parser.add_argument(
        '--opt_model_name',
        type=str,
        default='facebook/opt-2.7b',
        help='HuggingFace model name OR local folder path'
    )

    parser.add_argument('--opt_cache_dir', type=str, default=None)
    parser.add_argument('--alpha', type=float, default=0.5)

    parser.add_argument('--redis_ip', type=str, default='localhost')
    parser.add_argument('--redis_port', type=int, default=6379)
    parser.add_argument('--input_stream', type=str, default="remote_lm_input")
    parser.add_argument('--partial_output_stream', type=str, default="remote_lm_output_partial")
    parser.add_argument('--final_output_stream', type=str, default="remote_lm_output_final")

    args = parser.parse_args()
    main(args)
```

```python
%%writefile /kaggle/working/nejm-brain-to-text/model_training/data_augmentations.py
import torch
import torch.nn.functional as F
import numpy as np
from scipy.ndimage import gaussian_filter1d

def gauss_smooth(inputs, device, smooth_kernel_std=2, smooth_kernel_size=100, padding='same'):
    """
    Applies a 1D Gaussian smoothing operation with PyTorch to smooth the data along the time axis.
    Args:
        inputs (tensor : B x T x N): A 3D tensor with batch size B, time steps T, and number of features N.
                                     Assumed to already be on the correct device (e.g., GPU).
        smooth_kernel_std (float): Standard deviation of the Gaussian smoothing kernel.
        smooth_kernel_size (int): Size of the smoothing kernel before pruning.
        padding (str or int): Padding mode, either 'same' or 'valid' (or int for explicit padding).
        device (str): Device to use for computation (e.g., 'cuda' or 'cpu').
    Returns:
        smoothed (tensor : B x T x N): A smoothed 3D tensor with batch size B, time steps T, and number of features N.
    """
    # Get Gaussian kernel in numpy
    inp = np.zeros(smooth_kernel_size, dtype=np.float32)
    inp[smooth_kernel_size // 2] = 1
    gaussKernel = gaussian_filter1d(inp, smooth_kernel_std)
    validIdx = np.argwhere(gaussKernel > 0.01)
    gaussKernel = gaussKernel[validIdx]
    gaussKernel = np.squeeze(gaussKernel / np.sum(gaussKernel))

    # Convert to tensor (start float32)
    gaussKernel = torch.tensor(gaussKernel, dtype=torch.float32, device=device)
    gaussKernel = gaussKernel.view(1, 1, -1)  # [1, 1, kernel_size]

    # Prepare convolution
    B, T, C = inputs.shape
    inputs = inputs.permute(0, 2, 1)  # [B, C, T]
    gaussKernel = gaussKernel.repeat(C, 1, 1)  # [C, 1, kernel_size]

    # ---- PATCH: match kernel dtype/device to inputs ----
    gaussKernel = gaussKernel.to(device=inputs.device, dtype=inputs.dtype)
    # ---------------------------------------------------

    # Perform convolution
    smoothed = F.conv1d(inputs, gaussKernel, padding=padding, groups=C)
    return smoothed.permute(0, 2, 1)  # [B, T, C]
```

```python
%%writefile /kaggle/working/nejm-brain-to-text/language_model/language-model-standalone.py
#!/usr/bin/env python3
# import redis
import argparse
import numpy as np
from datetime import datetime
import time
import os
import re
import logging
import torch
import lm_decoder
from functools import lru_cache
from transformers import AutoModelForCausalLM, AutoTokenizer

# -------------------------------------------------------------------------
# Logging
# -------------------------------------------------------------------------
logging.basicConfig(
    format='%(asctime)s %(levelname)s: %(message)s',
    level=logging.INFO
)

# =========================================================================
# 1. Build n-gram decoder
# =========================================================================

def build_lm_decoder(
        model_path,
        max_active=7000,
        min_active=200,
        beam=17.,
        lattice_beam=8.0,
        acoustic_scale=1.5,
        ctc_blank_skip_threshold=1.0,
        length_penalty=0.0,
        nbest=1,
    ):

    decode_opts = lm_decoder.DecodeOptions(
        max_active,
        min_active,
        beam,
        lattice_beam,
        acoustic_scale,
        ctc_blank_skip_threshold,
        length_penalty,
        nbest
    )

    TLG_path = os.path.join(model_path, 'TLG.fst')
    words_path = os.path.join(model_path, 'words.txt')
    G_path = os.path.join(model_path, 'G.fst')
    rescore_G_path = os.path.join(model_path, 'G_no_prune.fst')

    if not os.path.exists(rescore_G_path):
        rescore_G_path = ""
        G_path = ""

    if not os.path.exists(TLG_path):
        raise ValueError(f'TLG file not found: {TLG_path}')

    if not os.path.exists(words_path):
        raise ValueError(f'words.txt not found: {words_path}')

    decode_resource = lm_decoder.DecodeResource(
        TLG_path,
        G_path,
        rescore_G_path,
        words_path,
        ""
    )

    decoder = lm_decoder.BrainSpeechDecoder(decode_resource, decode_opts)
    return decoder


# =========================================================================
# 2. Update n-gram decoder parameters
# =========================================================================

def update_ngram_params(
        ngramDecoder,
        max_active=200,
        min_active=17.0,
        beam=13.0,
        lattice_beam=8.0,
        acoustic_scale=1.5,
        ctc_blank_skip_threshold=1.0,
        length_penalty=0.0,
        nbest=100,
    ):
    
    decode_opts = lm_decoder.DecodeOptions(
        max_active,
        min_active,
        beam,
        lattice_beam,
        acoustic_scale,
        ctc_blank_skip_threshold,
        length_penalty,
        nbest,
    )

    ngramDecoder.SetOpt(decode_opts)


# =========================================================================
# 3. Build OPT & LLAMA model (PATCHED)
# =========================================================================

def build_opt(
        model_name='facebook/opt-2.7b',
        cache_dir=None,
        device='cuda' if torch.cuda.is_available() else 'cpu',
    ):
    """
    LOAD ANY OPT model:
        - HF model name (e.g., facebook/opt-2.7b)
        - OR LOCAL directory (e.g., /kaggle/working/opt_cache)

    This is the patched function.
    """

    logging.info(f"Loading OPT model from: {model_name}")
    logging.info(f"Cache dir: {cache_dir}")

    tokenizer = AutoTokenizer.from_pretrained(
        model_name,
        cache_dir=cache_dir
    )

    # fp16 for Kaggle (T4 supports fp16)
    model = AutoModelForCausalLM.from_pretrained(
        model_name,
        cache_dir=cache_dir,
        torch_dtype=torch.float16,
        low_cpu_mem_usage=True
    )

    if device != 'cpu':
        model = model.to(device)

    model.eval()

    # HF fix for OPT padding
    tokenizer.padding_side = "right"
    tokenizer.pad_token = tokenizer.eos_token

    logging.info(f"Loaded OPT model on device: {device}")
    return model, tokenizer


def build_tinyllama(model_name="TinyLlama/TinyLlama-1.1B-Chat-v1.0", cache_dir=None, device="cuda"):
    """
    Build TinyLlama HF model + tokenizer.
    We keep the interface parallel to build_opt so it can be used interchangeably.
    """
    tokenizer = AutoTokenizer.from_pretrained(model_name, cache_dir=cache_dir)
    model = AutoModelForCausalLM.from_pretrained(model_name, cache_dir=cache_dir)
    model.to(device)
    model.eval()
    return model, tokenizer

# =========================================================================
# 4. Rescoring using the OPT model
# =========================================================================

@torch.inference_mode()
def rescore_with_gpt2(model, tokenizer, device, hypotheses, length_penalty):

    model.eval()

    inputs = tokenizer(hypotheses, return_tensors='pt', padding=True)
    inputs = {k: v.to(device) for k, v in inputs.items()}

    outputs = model(**inputs)
    log_probs = torch.nn.functional.log_softmax(outputs.logits, dim=-1)
    log_probs = log_probs.cpu().numpy()

    input_ids = inputs['input_ids'].cpu().numpy()
    attention = inputs['attention_mask'].cpu().numpy()

    scores = []
    for i in range(len(hypotheses)):
        n_tokens = int(attention[i].sum())
        score = 0.0
        for t in range(1, n_tokens):
            score += log_probs[i, t-1, input_ids[i, t]]
        scores.append(score - length_penalty * n_tokens)

    return scores

def _compute_lm_logprob_per_token(model, tokenizer, sentences, device):
    """
    Compute average log probability per token for each sentence under a given HF LM.
    Returns: list of floats (one per sentence).
    """
    scores = []
    for s in sentences:
        # Encode with EOS if needed
        enc = tokenizer(
            s,
            return_tensors="pt",
            add_special_tokens=True
        ).to(device)

        with torch.no_grad():
            outputs = model(**enc)
            logits = outputs.logits  # [1, T, V]

        # Shift for next-token prediction
        shift_logits = logits[:, :-1, :]
        shift_labels = enc["input_ids"][:, 1:]

        # Cross-entropy (negative log likelihood)
        log_probs = torch.nn.functional.log_softmax(shift_logits, dim=-1)
        token_log_probs = log_probs.gather(-1, shift_labels.unsqueeze(-1)).squeeze(-1)

        # Average log prob per token
        avg_log_prob = token_log_probs.mean().item()
        scores.append(avg_log_prob)

    return scores

def ensemble_lm_decode(
    models,
    tokenizers,
    device,
    nbest,
    acoustic_scale=0.3,
    length_penalty=0.0,
    alpha=0.5,
):
    """
    Simple log-prob ensemble over multiple HF LMs.

    nbest: list of [sentence, ac_score, lm_score_from_ngram]
    models, tokenizers: lists of same length
    alpha: weight for LM scores vs acoustic (same spirit as gpt2_lm_decode)
    Returns: best sentence string
    """
    # Extract the sentences + acoustic scores
    sentences = [cand[0] for cand in nbest]
    ac_scores = [cand[1] for cand in nbest]

    # Compute LM scores for each model
    all_lm_scores = []
    for model, tok in zip(models, tokenizers):
        lm_scores = _compute_lm_logprob_per_token(model, tok, sentences, device)
        all_lm_scores.append(lm_scores)

    # Weighted ensemble: give more weight to OPT (index 0)
    opt_w = 0.8
    tiny_w = 0.2
    # Average LM log prob across models
    avg_lm_scores = []
    for i in range(len(sentences)):
        avg = opt_w * all_lm_scores[0][i] + tiny_w * all_lm_scores[1][i]
        avg_lm_scores.append(avg)

    # Combine acoustic + LM + length penalty
    best_idx = 0
    best_total = -1e9

    for i, sent in enumerate(sentences):
        # Lower length_penalty means favor shorter sentences if positive
        len_pen = length_penalty * len(sent.split())

        total_score = (
            acoustic_scale * ac_scores[i]   # from n-gram decoder
            + alpha * avg_lm_scores[i]      # ensemble LM
            - len_pen
        )

        if total_score > best_total:
            best_total = total_score
            best_idx = i

    return sentences[best_idx]


# =========================================================================
# 5. Decode with OPT model
# =========================================================================

def gpt2_lm_decode(
        model,
        tokenizer,
        device,
        nbest,
        acoustic_scale,
        length_penalty,
        alpha,
        returnConfidence=False,
        current_context_str=None,
    ):

    hypotheses = []
    ac_scores = []
    lm_scores = []

    # build hypothesis list
    for out in nbest:
        hyp = out[0].strip()
        if len(hyp) == 0:
            continue

        if current_context_str:
            hyp = current_context_str + " " + hyp
        
        hyp = hyp.replace('>', '')
        hyp = ' '.join(hyp.split())

        hypotheses.append(hyp)
        ac_scores.append(out[1])
        lm_scores.append(out[2])

    if len(hypotheses) == 0:
        return ("", []) if not returnConfidence else ("", [], 0.)

    ac_scores = np.array(ac_scores)
    lm_scores = np.array(lm_scores)

    # TRY full batch
    try:
        newLM = np.array(rescore_with_gpt2(model, tokenizer, device, hypotheses, length_penalty))

    except Exception as e:
        logging.error(f"Full OPT batch failed: {e}")
        newLM = []
        # fallback batches
        B = max(1, len(hypotheses)//5)
        for i in range(0, len(hypotheses), B):
            newLM.extend(
                rescore_with_gpt2(model, tokenizer, device, hypotheses[i:i+B], length_penalty)
            )
        newLM = np.array(newLM)

    # remove context prefix
    if current_context_str:
        prefix_len = len(current_context_str.split())
        hypotheses = [
            " ".join(h.split()[prefix_len:])
            for h in hypotheses
        ]

    totalScores = (
        acoustic_scale * ac_scores
        + (1 - alpha) * lm_scores
        + alpha * newLM
    )

    maxIdx = np.argmax(totalScores)
    best = hypotheses[maxIdx]

    # build nbest output
    nbest_out = []
    for i in range(len(hypotheses)):
        nbest_out.append(";".join(map(str, [
            hypotheses[i], ac_scores[i], lm_scores[i], newLM[i], totalScores[i]
        ])))

    if not returnConfidence:
        return best, nbest_out

    # softmax confidence
    t = totalScores - np.max(totalScores)
    p = np.exp(t)
    return best, nbest_out, p[maxIdx] / np.sum(p)


# =========================================================================
# 6. Utility functions
# =========================================================================

def connect_to_redis_server(redis_ip, redis_port):
    try:
        r = redis.Redis(host=redis_ip, port=redis_port)
        r.ping()
    except redis.exceptions.ConnectionError:
        logging.warning("Cannot connect to redis.")
        return None
    return r

def get_current_redis_time_ms(r):
    t = r.time()
    return int(t[0]*1000 + t[1]/1000)


# =========================================================================
# 7. Augment n-best list (unchanged)
# =========================================================================
# -- (UNCHANGED CODE OMITTED FOR BREVITY; KEEP YOUR EXISTING VERSION) --
# NOTE: You can keep your full augment_nbest and get_string_differences
# as-is. I removed it here just to keep the assistant message shorter.

# KEEP YOUR EXACT ORIGINAL FUNCTIONS HERE.
# =========================================================================


# =========================================================================
# 8. MAIN LOOP (unchanged except OPT loader patch)
# =========================================================================

def main(args):

    lm_path = args.lm_path
    gpu_number = args.gpu_number
    do_opt = args.do_opt
    opt_cache_dir = args.opt_cache_dir
    opt_model_name = args.opt_model_name     # <-- NEW ARGUMENT
    alpha = args.alpha

    lm_path = os.path.expanduser(lm_path)
    if opt_cache_dir:
        opt_cache_dir = os.path.expanduser(opt_cache_dir)

    # Pick GPU
    device = torch.device(f"cuda:{gpu_number}" if torch.cuda.is_available() else "cpu")
    logging.info(f"Using device: {device}")

    # ---------------------------------------------------------------------
    # Build OPT (patched)
    # ---------------------------------------------------------------------
    if do_opt:
        logging.info(f"Loading OPT model: {opt_model_name}")
        lm, lm_tokenizer = build_opt(
            model_name=opt_model_name,
            cache_dir=opt_cache_dir,
            device=device,
        )
        logging.info("OPT model loaded.")

    # ---------------------------------------------------------------------
    # Init n-gram decoder
    # ---------------------------------------------------------------------
    logging.info("Initializing n-gram decoder...")
    ngramDecoder = build_lm_decoder(
        lm_path,
        max_active=args.max_active,
        min_active=args.min_active,
        beam=args.beam,
        lattice_beam=args.lattice_beam,
        acoustic_scale=args.acoustic_scale,
        ctc_blank_skip_threshold=args.ctc_blank_skip_threshold,
        length_penalty=args.length_penalty,
        nbest=args.nbest,
    )

    # ---------------------------------------------------------------------
    # Connect to Redis
    # ---------------------------------------------------------------------
    r = None
    while r is None:
        r = connect_to_redis_server(args.redis_ip, args.redis_port)
        if r is None:
            logging.warning("Retrying redis connect in 3 seconds...")
            time.sleep(3)

    logging.info(f"Connected to Redis at {args.redis_ip}:{args.redis_port}")

    # ---------------------------------------------------------------------
    # Main decoding loop (UNCHANGED)
    # ---------------------------------------------------------------------
    # (KEEP YOUR ORIGINAL ENTIRE LOOP HERE — unchanged)
    #
    # I am omitting the huge loop here ONLY to keep the assistant message small.
    # Just keep your full original main loop and ONLY replace the LM-loading
    # section above.
    #
    # ---------------------------------------------------------------------


# =========================================================================
# 9. CLI ARGUMENTS  (PATCHED)
# =========================================================================

if __name__ == "__main__":

    parser = argparse.ArgumentParser()

    parser.add_argument('--lm_path', type=str, required=True)
    parser.add_argument('--gpu_number', type=int, default=0)

    # decoder hyperparams
    parser.add_argument('--max_active', type=int, default=7000)
    parser.add_argument('--min_active', type=int, default=200)
    parser.add_argument('--beam', type=float, default=17.0)
    parser.add_argument('--lattice_beam', type=float, default=8.0)
    parser.add_argument('--ctc_blank_skip_threshold', type=float, default=1.0)
    parser.add_argument('--length_penalty', type=float, default=0.0)
    parser.add_argument('--acoustic_scale', type=float, default=0.3)
    parser.add_argument('--nbest', type=int, default=100)
    parser.add_argument('--top_candidates_to_augment', type=int, default=20)
    parser.add_argument('--score_penalty_percent', type=float, default=0.01)
    parser.add_argument('--blank_penalty', type=float, default=9.0)

    parser.add_argument('--rescore', action='store_true')
    parser.add_argument('--do_opt', action='store_true')

    # PATCH: new OPT model selection
    parser.add_argument(
        '--opt_model_name',
        type=str,
        default='facebook/opt-2.7b',
        help='HuggingFace model name OR local folder path'
    )

    parser.add_argument('--opt_cache_dir', type=str, default=None)
    parser.add_argument('--alpha', type=float, default=0.5)

    parser.add_argument('--redis_ip', type=str, default='localhost')
    parser.add_argument('--redis_port', type=int, default=6379)
    parser.add_argument('--input_stream', type=str, default="remote_lm_input")
    parser.add_argument('--partial_output_stream', type=str, default="remote_lm_output_partial")
    parser.add_argument('--final_output_stream', type=str, default="remote_lm_output_final")

    args = parser.parse_args()
    main(args)
```
